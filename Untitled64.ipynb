```python
pdfreader = PdfReader("Light Functions Sequential Labeler Road User V1 Guideline.pdf")
```

```python
from typing_extensions import Concatenate

#read text from pdf
raw_text = ''
for i, page in enumerate(pdfreader.pages):
  content = page.extract_text()
  if content:
    raw_text += content
```

```python
raw_text
```

```python
import cassio
cassio.init(token=ASTRA_DB_APPLICATION_TOKEN, database_id=ASTRA_DB_ID)
```

Create the LangChain embedding and LLM objects for later usage:

```python
from langchain_google_genai import ChatGoogleGenerativeAI
llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash", google_api_key=API_KEY)
```

```python
embedding = GoogleGenerativeAIEmbeddings(model="gemini-embedding-001", google_api_key=API_KEY)
astra_vector_store = Cassandra(
    embedding=embedding,
    table_name="qa_mini_demo",
    session=None,
    keyspace=None
)
```

```python
from langchain_text_splitters import CharacterTextSplitter

text_splitter = CharacterTextSplitter(
    separator="\n",
    chunk_size=800,
    chunk_overlap=200,
    length_function = len,
)

texts = text_splitter.split_text(raw_text)
```

```python
texts[:50]
```

```python
astra_vector_store.add_texts(texts[:50])

print("Inserted %i headlines." % len(texts[:50]))

astra_vector_index = astra_vector_store.as_retriever()
```

Run the QA cycle
Simply run the cells and ask a question -- or quit to stop. (you can also stop execution with the "▪" button on the top toolbar)

Here are some suggested questions:

What is the current GDP?
How much the agriculture target will be increased to and what the focus will be

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough

# Define a simple prompt template for RAG
template = """Answer the question based only on the following context:
{context}

Question: {question}
"""
prompt = ChatPromptTemplate.from_template(template)

# Create a RAG chain
rag_chain = (
    {"context": astra_vector_index, "question": RunnablePassthrough()}
    | prompt
    | llm
    | StrOutputParser()
)

first_question = True
while True:
    if first_question:
        query_text = input("\nEnter your question (or type 'quit' to exit): ").strip()
    else:
        query_text = input("\nWhat's your next question (or type 'quit' to exit): ").strip()

    if query_text.lower() == "quit":
        break

    if query_text == "":
        continue

    first_question = False

    print("\nQUESTION: \"%s\"" % query_text)
    # Use the RAG chain to get the answer
    answer = rag_chain.invoke(query_text).strip()
    print("ANSWER: \"%s\"\n" % answer)

    print("FIRST DOCUMENTS BY RELEVANCE:")
    for doc, score in astra_vector_store.similarity_search_with_score(query_text, k=4):
        print("    [%0.4f] \"%s ...\"" % (score, doc.page_content[:84]))
```
